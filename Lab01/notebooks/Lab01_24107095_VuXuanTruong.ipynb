## Muc 10 - Cau hoi ly thuyet

1. **Agent la gi?** La thuc the dua ra quyet dinh (chon action) dua tren
   quan sat (observation) nhan duoc tu moi truong, nham toi da hoa tong
   reward tich luy theo thoi gian.

2. **Environment la gi?** La he thong ma agent tuong tac cung; nhan action
   tu agent, cap nhat trang thai noi bo va tra ve observation moi cung
   reward tuong ung.

3. **Observation khac action nhu the nao?** Observation la thong tin moi
   truong tra ve cho agent (dau vao cua agent), con action la quyet dinh
   agent gui cho moi truong (dau ra cua agent). Chung di nguoc chieu nhau
   trong vong lap tuong tac.

4. **Reward dung de lam gi?** La tin hieu so hoc (thuong la mot so thuc)
   ma moi truong tra ve sau moi buoc, dung de danh gia hanh dong agent vua
   thuc hien tot hay xau, va la muc tieu ma agent can toi da hoa ve lau
   dai.

5. **Episode la gi?** La mot chuoi tuong tac hoan chinh giua agent va moi
   truong, bat dau tu `reset()` va ket thuc khi `terminated` hoac
   `truncated` tra ve True.

6. **Policy la gi?** La chien luoc/ham anh xa tu observation (hoac state)
   sang action ma agent su dung de ra quyet dinh; co the la ham xac dinh,
   ngau nhien, hoac duoc hoc tu du lieu.

7. **`action_space` the hien dieu gi?** The hien tap hop tat ca cac action
   hop le ma agent co the thuc hien trong moi truong do (vi du: roi rac
   voi n action, hoac lien tuc voi khoang gia tri).

8. **`observation_space` the hien dieu gi?** The hien khong gian (cau
   truc, kieu du lieu, gioi han) cua moi observation ma moi truong co the
   tra ve cho agent.

9. **`reset()` duoc goi khi nao?** Duoc goi o dau moi episode (hoac khi
   can bat dau lai tu dau) de dua moi truong ve trang thai khoi tao va
   nhan observation dau tien.

10. **`step(action)` thuc hien viec gi?** Thuc hien mot action trong moi
    truong, cap nhat trang thai noi bo, va tra ve
    `(observation, reward, terminated, truncated, info)` tuong ung.

11. **`terminated` va `truncated` khac nhau the nao?** `terminated` la
    ket thuc "tu nhien" theo dinh nghia bai toan (thang/thua/xong viec);
    `truncated` la bi cat ngang boi gioi han ben ngoai (vi du time limit),
    khong phan anh ket qua thanh cong hay that bai cua agent.

12. **Random policy co phai la mot thuat toan hoc tang cuong khong?**
    Khong. Random policy khong hoc gi tu du lieu/kinh nghiem, no chi chon
    action ngau nhien bat ke observation. No thuong duoc dung lam baseline
    (moc so sanh) de danh gia cac policy/thuat toan hoc duoc co thuc su
    tot hon hanh dong ngau nhien hay khong.

13. **Vi sao phai chay nhieu episode thay vi chi mot episode?** Vi moi
    truong va policy co the co tinh ngau nhien (do action ngau nhien,
    moi truong stochastic, seed khac nhau...), nen ket qua cua mot episode
    don le co the khong dai dien. Chay nhieu episode giup uoc luong dang
    tin cay hon (mean, std) ve hieu suat trung binh cua policy.

14. **Vi sao can random seed?** De dam bao tinh tai lap cua thi nghiem:
    cung mot seed se cho cung mot chuoi ngau nhien (observation ban dau,
    action mau...), giup so sanh cong bang giua cac lan chay/policy khac
    nhau va giup nguoi khac kiem tra lai ket qua.

15. **Reward trung binh co y nghia gi khi danh gia agent?** Reward trung
    binh (mean reward) qua nhieu episode phan anh hieu suat tong the/on
    dinh cua agent trong dai han, giam bot anh huong cua cac episode ca
    biet (may man hoac xui xeo), va la thuoc do pho bien de so sanh cac
    agent/policy khac nhau.
